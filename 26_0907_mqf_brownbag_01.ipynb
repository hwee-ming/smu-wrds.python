```python
# Apple Inc's Stock Splits 4-1, 7-1, 2-1, 2-1, and 2-1
query = """
          SELECT dlycaldt     AS "Date",
                 dlyfacprc    AS "Price Adj. Factor",
               --dlycumfacpr  AS "Cumulative Price Adj. Factor",
                 dlycumfacshr AS "Cumulative Share Adj. Factor",
                 prev_dlycumfacshr/dlycumfacshr AS "prev_dlycumfacshr/dlycumfacshr"
            FROM (
                  SELECT dlycaldt,
                         dlyfacprc,
                       --dlycumfacpr,
                         dlycumfacshr,
                         LAG(dlycumfacshr) OVER (PARTITION BY permno ORDER BY dlycaldt) AS "prev_dlycumfacshr"
                    FROM crsp_a_stock.dsf_v2
                   WHERE permno = 14593
                 ) sub
           WHERE dlycumfacshr IS DISTINCT FROM prev_dlycumfacshr
        ORDER BY dlycaldt
        """
apple_stock_split = db.raw_sql(query)
apple_stock_split
```

```python
# LogicMark Inc's Reverse Stock Splits 1-25, 1-20, 1-10, and 1-10
query = """
          SELECT dlycaldt     AS "Date",
                 dlyfacprc    AS "Price Adj. Factor",
               --dlycumfacpr  AS "Cumulative Price Adj. Factor",
                 dlycumfacshr AS "Cumulative Share Adj. Factor",
                 prev_dlycumfacshr/dlycumfacshr AS "prev_dlycumfacshr/dlycumfacshr"
            FROM (
                  SELECT dlycaldt,
                         dlyfacprc,
                       --dlycumfacpr,
                         dlycumfacshr,
                         LAG(dlycumfacshr) OVER (PARTITION BY permno ORDER BY dlycaldt) AS "prev_dlycumfacshr"
                    FROM crsp_a_stock.dsf_v2
                   WHERE permno = 14934
                 ) sub
           WHERE dlycumfacshr IS DISTINCT FROM prev_dlycumfacshr
        ORDER BY dlycaldt
        """
logicmark_reverse_split = db.raw_sql(query)
logicmark_reverse_split
```

```python
# Barnes & Noble Education Inc's 1-for-100 Reverse Stock Split
# Prc / Cumulative Prc Adj Factor
# Shrout * Cumulative Shr ADj Factor
BNED = db.raw_sql("""SELECT --permno,
                            --SUBSTRING(dlycaldt::TEXT, 1, 4) AS year,
                              dlycaldt,
                              dlyprc,
                            --dlyfacprc,
                              dlycumfacpr,
                              dlyprc/dlycumfacpr   AS "dlyprc/dlycumfacpr",
                              shrout,
                              dlycumfacshr,
                              shrout*dlycumfacshr  AS "shrout*dlycumfacshr"
                       FROM   crsp_a_stock.dsf_v2
                      WHERE   permno = 15489
                        AND   dlycaldt >= '2024-06-07'
                        AND   dlycaldt <= '2024-06-17'
                   ORDER BY   1
                  """)
BNED
```

```python
# CRSP’s S&P 500 Index Constituents – Dual Class Shares on Last Trading Day of 2025
query = """
               SELECT a.permco, a.permno, a.shareclass, a.ticker
                 FROM crsp_a_stock.wrds_dsfv2_query a
           INNER JOIN crsp_a_indexes.dsp500list_v2  b
                   ON a.permno = b.permno
                WHERE a.dlycaldt = '12/31/2025'
                  AND b.mbrenddt = '12/31/2025'
             GROUP BY a.permco, a.permno, a.shareclass, a.ticker
               HAVING a.permco IN (
                                      SELECT c.permco
                                        FROM crsp_a_stock.wrds_dsfv2_query c
                                  INNER JOIN crsp_a_indexes.dsp500list_v2  d
                                          ON c.permno = d.permno
                                       WHERE c.dlycaldt = '12/31/2025'
                                         AND d.mbrenddt = '12/31/2025'
                                    GROUP BY c.permco
                                      HAVING COUNT(DISTINCT c.shareclass) > '1'
                                  )
             ORDER BY a.permco, a.shareclass
        """
sp500_dual_class_share = db.raw_sql(query)
sp500_dual_class_share
```

```python
# TESLA's S&P 500 debut 21 Dec 2020
tsla = db.raw_sql("""
    SELECT
          DATE_TRUNC('second', time_m::TIME)                   AS second_interval,
          SUM(size)                                            AS tradevol_sum,
          AVG(price)                                           AS price_avg,
          PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY price)   AS price_median,
          MIN(price)                                           AS price_min,
          MAX(price)                                           AS price_max
    FROM taqm_2020.ctm_20201218
    WHERE sym_root = 'TSLA'
      AND time_m >= '15:55:00'
      AND time_m <= '16:00:00'
    GROUP BY second_interval
    ORDER BY second_interval
""", index_col=['second_interval'])
tsla
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import pandas as pd

# Ensure index sorted (15:55 → 16:00)
tsla = tsla.sort_index()

# --- Handle TimedeltaIndex vs DatetimeIndex ---
if isinstance(tsla.index, pd.TimedeltaIndex):
    # Anchor to 18 Dec 2020
    tsla.index = pd.to_datetime('2020-12-18') + tsla.index
else:
    # Already datetime — normalize date to 18 Dec 2020
    tsla.index = tsla.index.map(lambda t: t.replace(year=2020, month=12, day=18))

# --- Plot setup ---
fig, ax1 = plt.subplots(figsize=(12, 6))

# --- Left axis: Trade volume ---
ax1.bar(tsla.index, tsla['tradevol_sum'],
        label='Trade Volume (sum)',
        color='lightgray', alpha=0.6,
        width=pd.Timedelta(seconds=1))  # width matches 1 second
ax1.set_ylabel('Trade Volume (shares)')
ax1.tick_params(axis='y')

# --- Right axis: Prices ---
ax2 = ax1.twinx()

# Shaded min–max range
ax2.fill_between(tsla.index, tsla['price_min'], tsla['price_max'],
                 color='purple', alpha=0.1, label='Price Range (Min–Max)')

# Line plots
ax2.plot(tsla.index, tsla['price_avg'], label='Avg Price', color='blue', marker='o', linewidth=1.5)
ax2.plot(tsla.index, tsla['price_median'], label='Median Price', color='green', linestyle='--', marker='s', linewidth=1.5)
ax2.plot(tsla.index, tsla['price_min'], label='Min Price', color='red', linestyle=':', marker='x', linewidth=1.2)
ax2.plot(tsla.index, tsla['price_max'], label='Max Price', color='purple', linestyle=':', marker='x', linewidth=1.2)

ax2.set_ylabel('Trade Price (USD)')
ax2.tick_params(axis='y')

# --- Title, labels, ticks ---
plt.title('TSLA Trade Activity\n18 Dec 2020, 15:55–16:00 (Last 5 Minutes Before Close)')
ax1.set_xlabel('Time (HH:MM:SS)')

# Format x-axis as HH:MM:SS with minute ticks only
ax1.xaxis.set_major_formatter(mdates.DateFormatter("%H:%M:%S"))
ax1.xaxis.set_major_locator(mdates.MinuteLocator())
fig.autofmt_xdate()

plt.grid(True, linestyle='--', alpha=0.6)

# --- Legend outside the plotting box, shifted right ---
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines1 + lines2, labels1 + labels2,
           loc='upper left', bbox_to_anchor=(1.15, 1),
           borderaxespad=0, frameon=False)

plt.tight_layout()
plt.show()

```
